    Some values are missing and need to be handled.
    
    Categorical features (Make, Colour) must be transformed into a numerical format for machine learning models.

    Drop rows with excessive missing values if necessary
    
    Encoding Categorical Features:
    Apply OneHotEncoder from sklearn.preprocessing to convert Make and Colour into numerical format.

```python
# Importing Libraries
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_absolute_error, mean_squared_error

```

```python
# Getting The Data 
data = pd.read_csv("./Csv Files/car-sales-data.csv")
```

```python
#Checking For missing values
data.isna().sum()
```

```python
#Dropping the null price Column 
data.dropna(subset=['Price'],inplace=True)
```

```python
#filling in the missing values of other features using SimpleImpputer from Scikit Learn
category_features = ['Make','Colour']
door_features = ['Doors']
numerical_features = ['Odometer (KM)']

category_imputer = SimpleImputer(strategy="constant",fill_value="missing")
door_imputer = SimpleImputer(strategy="constant",fill_value=4)
numerical_imputer = SimpleImputer(strategy="mean")

imputer = ColumnTransformer(
    [("cat_imputer",category_imputer,category_features),
    ("door_imputer",door_imputer,door_features),
    ("num_imputer",numerical_imputer,numerical_features)]
,remainder="passthrough")

filled_data = imputer.fit_transform(data)
```

```python
#splitting the data into inputs and labels
filled_data = pd.DataFrame(data=filled_data,
                          columns=["Make","Colour","Doors","Odometer (KM)","Price"])

X = filled_data.drop("Price",axis=1)
y = filled_data['Price']
```

```python
#Transforming the categorical features into numerical with OneHotEncoder from Sklearn Preprocessing 
one_hot = OneHotEncoder()
tranformer = ColumnTransformer([("oneHot",one_hot,category_features)],remainder="passthrough")
transformer_X = transformer.fit_transform(X)
```

```python
transformer_X
```

```python
transformed_X = pd.DataFrame(transformer_X)
```

```python
transformed_X
```

# Model Selection And Training

```python
np.random.seed(42)

#Splitting the data into Training set and Test set using Train test split fromo sklearn.model_selection
Xtrain , Xtest , ytrain , ytest = train_test_split(transformed_X,y,test_size=0.3)
Xtrain
```

```python
#Training LinearRegression on Xtrain
model = LinearRegression()
model.fit(Xtrain,ytrain)
model.score(Xtest,ytest)
```

```python
#Training on different algorithms
regression_models ={
    "Linear Regression":LinearRegression(),
    "Random Forest Regressor": RandomForestRegressor(),
    "Gradient Boosting Regressor": GradientBoostingRegressor()
}

#Define a function  fit and score to train on different model
def fit_and_score(models,training_setX,training_setY,test_setX,test_setY):
    model_scores = {}
    np.random.seed(42)
    for name,model in models.items():
        model.fit(training_setX,training_setY) 
        model_scores[name] = model.score(test_setX,test_setY)
    return model_scores
```

```python
model_scores = fit_and_score(models=regression_models,
                 training_setX=Xtrain,
                 training_setY =  ytrain,
                 test_setX=Xtest,
                 test_setY=ytest)
```

```python
model_compare = pd.DataFrame(data = model_scores,index=["Accuracy"])
```

```python
model_compare.T.plot(kind="bar")
```

     After training the data on LinearRegression , RandomForestRegressor , GradientBoostingRegressor it can be observed that GradientBoostingRegressor have  yielded the best results 

### Hyper tuning our Machine Learning model with different Parameters Using GridSearchCv from sklearn.model_selection

```python
gdr_model = GradientBoostingRegressor()

gdr_parameters= {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.05, 0.1, 0.2],
    'max_depth': [3, 5, 7],
    'subsample': [0.7, 0.8, 0.9]
}

grid_search = GridSearchCV(gdr_model, param_grid =gdr_parameters , cv = 5,n_jobs=-1)
grid_search.fit(Xtrain,ytrain)
```

```python
best_params = grid_search.best_params_
print(best_params)
```

```python
np.random.seed(42)

gdr_model = GradientBoostingRegressor(learning_rate=0.05,
                                        max_depth=3,
                                        n_estimators=100,
                                     subsample=0.9)
```

```python
gdr_model.fit(Xtrain,ytrain)
gdr_model.score(Xtest,ytest)
```

### Model Evalutaion 
    Evaluting the model on cross_val_score and different scoring metrics for better generalization on unseen data and to prevent overfitting

```python
#Defing a fucntion for different scoring parameters
metrics = {
    "r2 Score": "r2",
    "Mean absolute Error": "neg_mean_absolute_error",
    "Mean Squared Error" : "neg_mean_squared_error"
}

def cross_val_scores(model,X,y):
    metric_scores={}

    for name,metric in metrics.items():
        score = cross_val_score(model,X,y,scoring=metric)
        metric_scores[name] = score.mean()
    return metric_scores
```

```python
metric_scores = cross_val_scores(model=gdr_model,X=transformed_X,y=y)
```
